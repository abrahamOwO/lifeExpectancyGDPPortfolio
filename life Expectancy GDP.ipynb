```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("Life-Expectancy-and-GDP-Starter/all_data.csv")
df["Country"] = df["Country"].replace({"United States of America":"United States"})
print(df.head())
print(df.info)
def get_zip_data():
    return zip(
    range(len(df["Country"].unique())),
    df["Country"].unique(),
    ["purple","grey","orange","green","blue","brown"])
zip_data = get_zip_data
```

```python
sns.lmplot(data=df, x="Year", y="Life expectancy at birth (years)", hue="Country")
plt.show()
plt.clf()
```

```python
fig,axs=plt.subplots(nrows=2,ncols=3,figsize=(12,7))
axs=axs.ravel()
plt.subplots_adjust(wspace=0.3, hspace=0.3, top=0.9)
tick_marks = range(0,2)


for i,country,color in zip_data:
    country_df = df[df["Country"]==country]
    axN = axs[i]
    axN.set(ylim=(40,85))
    axN.set_title(country)
    # axN.set_xticks(tick_marks)
    # axN.set_xticklabels(tick_marks)
    sns.regplot(data=country_df,x="GDP",y="Life expectancy at birth (years)",ax=axN,color=color,line_kws=dict(color="r",linewidth="1"))
    axN.set_ylabel("Life Exp")
    axN.set_xlabel("GDP")

    # model = LinearRegression()
    # model.fit(X=country_df.GDP, y=country_df["Life expectancy at birth (years)"])
    # y_prediction = model.predict(country_df["GDP"])
    # axN.plot(country_df["GDP"], y_prediction)

# plt.show()
# plt.clf()
# figure = plt.figure(figsize=(8,2))
# plt.figlegend([country for country in df["Country"].unique()],loc=6)
fig.suptitle("Life Expectancy against GDP")
plt.show()
plt.clf()
```
